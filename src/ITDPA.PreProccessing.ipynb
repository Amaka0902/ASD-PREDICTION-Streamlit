```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer


# Load the dataset
df = pd.read_csv(r'train.csv')


# Drop the 'ethnicity' column
if 'ethnicity' in df.columns:
    df = df.drop(columns=['ethnicity'])

# Preview the data
print("Data Shape after dropping 'ethnicity':", df.shape)
print("First 5 rows:\n", df.head())

# Identify target column (replace 'target' with your actual target column name)
target_column = 'target'  # Replace with your actual target column name
X = df.drop(columns=[target_column])
y = df[target_column]

# Separate columns by data type
categorical_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Define preprocessing for numerical features
numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

# Define preprocessing for categorical features
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Combine transformers
preprocessor = ColumnTransformer(transformers=[
    ('num', numerical_transformer, numerical_cols),
    ('cat', categorical_transformer, categorical_cols)
])

# Apply transformations
X_processed = preprocessor.fit_transform(X)

# Output shapes
print("Processed Features Shape:", X_processed.shape)
print("Target Shape:", y.shape)

# Load the dataset
df = pd.read_csv(r'train.csv')

# Drop the 'ethnicity' column if it exists
if 'ethnicity' in df.columns:
    df = df.drop(columns=['ethnicity'])

# Save the cleaned dataset to a new CSV file
df.to_csv(r'cleaned_train.csv', index=False)

# Preview the cleaned data
print("Data Shape after dropping 'ethnicity':", df.shape)
print("First 5 rows:\n", df.head())

```
