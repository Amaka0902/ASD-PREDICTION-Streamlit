```python
# ASD Dataset Cleaning, Prediction, and Export Script (Zahra Version - Correct Path)
# -------------------------------------------------------------------

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

# 1. Load your dataset (CSV file)
df = pd.read_csv(r"C:\Users\Zahra\OneDrive\Documents\ITDPA QUESTION 4 ASSIGNMENT\train.csv")

# 2. Drop duplicates and handle missing values
df = df.drop_duplicates()
df = df.dropna()

# 3. Encode categorical columns if needed
label_encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    if col.lower() not in ['class', 'predicted_class']:  # skip target columns
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col])
        label_encoders[col] = le

# 4. Define features (X) and target (y)
# If your dataset's target column has a different name, change 'Class' below
target_column = 'Class' if 'Class' in df.columns else df.columns[-1]
X = df.drop(columns=[target_column], errors='ignore')
y = df[target_column]

# 5. Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 6. Train a Random Forest model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# 7. Make predictions and get probabilities
y_pred = model.predict(X)
y_prob = model.predict_proba(X)[:, 1]  # probability of ASD

# 8. Add predictions and probabilities to the dataset
df['Predicted_Class'] = y_pred
df['Probability'] = y_prob.round(3)

# 9. Save the cleaned dataset with predictions
output_path = r"C:\Users\Zahra\OneDrive\Documents\ITDPA QUESTION 4 ASSIGNMENT\final_cleaned_asd_dataset.xlsx"
df.to_excel(output_path, index=False)

print("✅ Dataset cleaned, predicted, and saved as:")
print(output_path)

# 10. Optional: model evaluation
print("\nClassification Report:")
print(classification_report(y_test, model.predict(X_test)))
```
